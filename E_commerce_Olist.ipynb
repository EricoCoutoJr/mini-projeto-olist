## **Sprint 1: Configuração e Ingestão**

Configurar o ambiente de desenvolvimento, criar o repositório (se ainda não o fez) e implementar a lógica para ler os arquivos CSV ('olist_products_dataset.csv' e 'olist_orders_dataset.csv') utilizando o módulo nativo `csv.DictReader`, garantindo que os dados sejam carregados em estruturas de dados Python adequadas (listas de dicionários, por exemplo).


**Descrição**:
As quatro primeiras instruções envolvem importar o módulo `csv`, definir caminhos de arquivo, implementar uma função para ler arquivos CSV e convertê-los em uma lista de dicionários, e chamar essa função para ambos os conjuntos de dados. Essas etapas são interdependentes e podem ser executadas de forma eficiente em um único bloco de código para configurar a ingestão dos dados. A quinta instrução, que exibe as primeiras entradas, também foi incluída para verificar imediatamente se a ingestão de ambos os conjuntos de dados ocorreu com sucesso.



```python
import csv
import os

# Define os caminhos dos arquivos
products_file_path = 'olist_products_dataset.csv'
orders_file_path = 'olist_orders_dataset.csv'

# Função para ler CSV usando csv.DictReader
def read_csv_to_dict(file_path):
    data = []
    try:
        with open(file_path, mode='r', encoding='utf-8') as file:
            reader = csv.DictReader(file)
            for row in reader:
                data.append(row)
        print(f"Dados carregados com sucesso: {len(data)} linhas de {file_path}")
    except FileNotFoundError:
        print(f"Erro: O arquivo {file_path} não foi encontrado.")
    except Exception as e:
        print(f"Ocorreu um erro ao ler {file_path}: {e}")
    return data

# Carrega os datasets
products_data = read_csv_to_dict(products_file_path)
orders_data = read_csv_to_dict(orders_file_path)

# Exibe as primeiras 5 entradas de products_data para verificação
print("\nPrimeiras 5 entradas de products_data:")
for i, row in enumerate(products_data[:5]):
    print(f"Linha {i+1}: {row}")

# Exibe as primeiras 5 entradas de orders_data para verificação
print("\nPrimeiras 5 entradas de orders_data:")
for i, row in enumerate(orders_data[:5]):
    print(f"Linha {i+1}: {row}")
```

## **Sprint 2: Transformação de Strings**

Desenvolver funções Python para padronizar os nomes das categorias de produtos na coluna 'product_category_name'. Isso inclui converter todas as strings para letras minúsculas (`.lower()`), remover espaços em branco excedentes no início e no fim (`.strip()`) e utilizar Expressões Regulares (`re`) para limpar caracteres especiais ou pontuações indevidas.


**Descrição**:
Implementa a função de limpeza e aplique-a às categorias de produtos conforme as instruções; em seguida, exiba as 5 primeiras entradas para conferir.



```python
import re

def clean_category_name(category_name):
    if category_name is None:
        return None
    # Converte a string para minúsculas
    cleaned_name = category_name.lower()
    # Remove espaços em branco extras no início e no fim
    cleaned_name = cleaned_name.strip()
    # Usa regex para remover caracteres especiais/pontuação, mantendo apenas alfanuméricos e espaços
    cleaned_name = re.sub(r'[^a-z0-9 ]', '', cleaned_name)
    return cleaned_name

# Aplica a função de limpeza à coluna 'product_category_name' em products_data
for product in products_data:
    product['product_category_name'] = clean_category_name(product.get('product_category_name'))

# Exibe as primeiras 5 entradas de products_data para verificar as alterações
print("\nPrimeiras 5 entradas de products_data após a limpeza dos nomes das categorias:")
for i, row in enumerate(products_data[:5]):
    print(f"Linha {i+1}: {row}")
```

## **Sprint 3: Limpeza de Nulos nos Produtos**

Implemente a lógica para tratar valores ausentes no conjunto de dados de produtos. Isso inclui preencher categorias nulas e lidar com dimensões físicas nulas.


**Descrição**:
Preciso inicializar um dicionário para armazenar as contagens de valores nulos para fins de verificação. Em seguida, iterarei sobre `products_data` para preencher valores nulos na coluna `product_category_name` e converter as colunas de dimensões físicas para o formato numérico, tratando possíveis erros. Após a conversão, calcularei e utilizarei a média de cada dimensão física para preencher os valores nulos restantes. Por fim, exibirei as contagens de valores nulos antes e depois do processamento.



**A justificativa para usar a mediana em vez da média para preencher valores ausentes em dimensões físicas de produtos reside na sua robustez estatística, especialmente na presença de outliers.**

**Por que a Mediana é Mais Robusta?**

1.  **Resistência a Outliers:** Em dados de dimensões físicas, como peso, comprimento, altura e largura de produtos, é comum encontrar valores extremos (outliers). Por exemplo, um produto muito pequeno ou excepcionalmente grande pode distorcer significativamente a média. A média é calculada somando todos os valores e dividindo pelo número de valores, o que a torna altamente sensível a valores muito altos ou muito baixos. A mediana, por outro lado, é o valor do meio em um conjunto de dados ordenado. Ela não é afetada pelos valores exatos dos outliers, apenas pela sua posição.

2.  **Representatividade:** Ao minimizar o impacto de valores extremos, a mediana proporciona uma estimativa mais representativa do valor "típico" ou "central" dos dados. Isso significa que, ao imputar valores ausentes com a mediana, estamos usando um valor que reflete melhor a maioria dos produtos, evitando que a imputação seja "puxada" por alguns poucos valores atípicos.

3.  **Evita Distorções:** Em contextos de machine learning e análise de dados, a distorção da distribuição original dos dados devido a imputação inadequada pode levar a modelos enviesados ou com menor poder preditivo. Usar a mediana ajuda a manter a forma geral da distribuição dos dados, o que é crucial para muitos algoritmos.

Em resumo, a mediana é uma escolha superior para imputação de valores ausentes em dimensões físicas porque oferece uma medida de tendência central que é menos suscetível à influência de outliers, resultando em uma imputação mais confiável e menos propensa a introduzir viés ou distorcer a análise subsequente.

```python
import numpy as np

# Inicializa um dicionário para armazenar a contagem de nulos antes do tratamento
initial_null_counts = {
    'product_category_name': 0,
    'product_weight_g': 0,
    'product_length_cm': 0,
    'product_height_cm': 0,
    'product_width_cm': 0
}

# Conta os nulos iniciais
for product in products_data:
    if product.get('product_category_name') is None or product.get('product_category_name') == '':
        initial_null_counts['product_category_name'] += 1
    for dim_col in ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']:
        if product.get(dim_col) is None or product.get(dim_col) == '':
            initial_null_counts[dim_col] += 1

print("Contagem de Nulos Iniciais:", initial_null_counts)

# Passo 1: Preencher valores nulos em 'product_category_name' com 'sem categoria'
for product in products_data:
    if product.get('product_category_name') is None or product.get('product_category_name') == '':
        product['product_category_name'] = 'sem categoria'

# Passo 2 e 3: Converter dimensões físicas para numérico e preencher nulos com a mediana
# Coletar todos os valores numéricos para cálculo da mediana
weight_values = []
length_values = []
height_values = []
width_values = []

for product in products_data:
    for dim_col, value_list in [
        ('product_weight_g', weight_values),
        ('product_length_cm', length_values),
        ('product_height_cm', height_values),
        ('product_width_cm', width_values)
    ]:
        val = product.get(dim_col)
        if val is not None and val != '':
            try:
                num_val = float(val)
                product[dim_col] = num_val
                value_list.append(num_val)
            except ValueError:
                product[dim_col] = None # Definir como None se a conversão falhar
        else:
            product[dim_col] = None # Definir como None se estiver vazio ou None

# Calcular as medianas para cada dimensão, lidando com casos onde a lista pode estar vazia
median_weight = np.median(weight_values) if weight_values else 0
median_length = np.median(length_values) if length_values else 0
median_height = np.median(height_values) if height_values else 0
median_width = np.median(width_values) if width_values else 0

# Justificativa para usar a mediana: A mediana é uma medida de tendência central mais robusta do que a média em presença de outliers, que são comuns em dados de dimensões físicas de produtos (por exemplo, produtos muito pequenos ou muito grandes). Ela minimiza o impacto de valores extremos, proporcionando uma estimativa mais representativa para preenchimento de dados ausentes e evitando distorções na distribuição original.
for product in products_data:
    if product.get('product_weight_g') is None:
        product['product_weight_g'] = median_weight
    if product.get('product_length_cm') is None:
        product['product_length_cm'] = median_length
    if product.get('product_height_cm') is None:
        product['product_height_cm'] = median_height
    if product.get('product_width_cm') is None:
        product['product_width_cm'] = median_width

# Inicializa um dicionário para armazenar a contagem de nulos após o tratamento
final_null_counts = {
    'product_category_name': 0,
    'product_weight_g': 0,
    'product_length_cm': 0,
    'product_height_cm': 0,
    'product_width_cm': 0
}

# Conta os nulos finais
for product in products_data:
    if product.get('product_category_name') is None or product.get('product_category_name') == '':
        final_null_counts['product_category_name'] += 1
    for dim_col in ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']:
        if product.get(dim_col) is None:
            final_null_counts[dim_col] += 1

print("Contagem de Nulos Finais:", final_null_counts)

# Exibir as primeiras 5 entradas de products_data para verificar as alterações
print("\nPrimeiras 5 entradas de products_data após o tratamento de nulos:")
for i, row in enumerate(products_data[:5]):
    print(f"Linha {i+1}: {row}")
```

## **Sprint 4: Regras de Negócio e Formatação de Datas**

No dataset de pedidos, implementar a validação da regra de negócio: verificar se as datas de entrega nulas ('order_delivered_customer_date') correspondem obrigatoriamente a pedidos com status 'canceled'. Além disso, utilizar o módulo `datetime` para converter a coluna 'order_approved_at' do formato original de string para o formato de data simplificado brasileiro 'DD/MM/YYYY'.


**Descrição**:
Implementa a validação de regras de negócio e a formatação de data para `orders_data` conforme as instruções e, em seguida, exiba os 5 primeiros registros para verificar as alterações.



```python
from datetime import datetime

# Contador de violações de regras de negócio
business_rule_violations = 0

# 1. Iterar sobre orders_data para aplicar validação da regra de negócio e formatar datas
for order in orders_data:
    # Validação da regra de negócio: 'order_delivered_customer_date' nulo deve corresponder ao status 'canceled'
    delivered_customer_date = order.get('order_delivered_customer_date')
    order_status = order.get('order_status')

    if (delivered_customer_date is None or delivered_customer_date == '') and order_status != 'canceled':
        print(f"AVISO: Violação da regra de negócio para o ID do pedido {order.get('order_id')}: "
              f"'order_delivered_customer_date' é nulo, mas o 'order_status' é '{order_status}'.")
        business_rule_violations += 1

    # Converter 'order_approved_at' para o formato 'DD/MM/YYYY'
    approved_at_str = order.get('order_approved_at')
    if approved_at_str:
        try:
            # Analisar a string original em um objeto datetime
            dt_object = datetime.strptime(approved_at_str, '%Y-%m-%d %H:%M:%S')
            # Formatar o objeto datetime para 'DD/MM/YYYY'
            order['order_approved_at'] = dt_object.strftime('%d/%m/%Y')
        except ValueError:
            # Lidar com casos onde o formato da data pode ser inesperado
            print(f"Não foi possível analisar 'order_approved_at': {approved_at_str} para o ID do pedido {order.get('order_id')}")
            order['order_approved_at'] = None # Ou manter o original, dependendo do tratamento de erro desejado
    else:
        order['order_approved_at'] = None # Garantir None consistente para valores verdadeiramente nulos

print(f"\nTotal de violações de regras de negócio encontradas: {business_rule_violations}")

# 3. Exibir as primeiras 5 entradas de orders_data para verificar as alterações
print("\nPrimeiras 5 entradas de orders_data após a formatação de datas e verificação de regras:")
for i, row in enumerate(orders_data[:5]):
    print(f"Linha {i+1}: {row}")
```
